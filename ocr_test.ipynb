```python
import cv2
import numpy as np
```

```python
img_path = [r'E:\AiLib_ObjectDetection\image_crop\0.jpg',
            r'E:\AiLib_ObjectDetection\image_crop\1.jpg',
            r'E:\AiLib_ObjectDetection\image_crop\2.jpg',
            r'E:\AiLib_ObjectDetection\image_crop\3.jpg',
            r'E:\AiLib_ObjectDetection\image_crop\4.jpg',
            r'E:\AiLib_ObjectDetection\image_crop\5.jpg',
            r'E:\AiLib_ObjectDetection\image_crop\6.jpg',
            r'E:\AiLib_ObjectDetection\image_crop\7.jpg',
            r'E:\AiLib_ObjectDetection\image_crop\8.jpg',
            r'E:\AiLib_ObjectDetection\image_crop\9.jpg',
            ]
imgs = []
for i,v in enumerate(img_path): 
    imgs.append(cv2.imread(v))
```

### Image Processing 

```python
def resize_for_ocr(img, target_height=1000):
    h, w = img.shape[:2]
    scale = target_height / h
    new_w = int(w * scale)
    resized = cv2.resize(img, (new_w, target_height), interpolation=cv2.INTER_LINEAR)
    return resized

resized = resize_for_ocr(imgs[0], target_height=1000)
# cv2.imshow("resized", resized)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
```

```python
kernel_size_row = 3
kernel_size_col = 3
kernel = np.ones((3, 3), np.uint8)
# dilation = cv2.dilate(edges, kernel, iterations=1)
```

```python
gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)

# 블러로 노이즈 제거
blur = cv2.GaussianBlur(gray, (5, 5), 0)
# 엣지 검출 (Canny)

edges = cv2.Canny(blur, 0, 15, apertureSize=3)

dilation = cv2.dilate(edges, kernel, iterations=1)
erode1 = cv2.erode(dilation, kernel, iterations = 1)
```

```python
# cv2.imshow("g",gray)
cv2.imshow("d",dilation)
cv2.imshow("e",erode1) 
cv2.waitKey(0)
cv2.destroyAllWindows()
```

EasyOCR

```python
import easyocr
reader = easyocr.Reader(['ko','en'], gpu = False) # 배포시에는 True

result = reader.readtext(erode1, detail=1, paragraph=False)

result
# _, text, conf = result[0]
# print(text.strip, float(conf))
```
